```python
import pandas as pd
import numpy as np

# 1. Referans genleri yükle
periodic_path = "/content/spellman_periodic_genes.txt"
periodic_df = pd.read_csv(periodic_path, header=None, names=["ORF"])
periodic_genes = set(periodic_df["ORF"].str.strip().tolist())
print(f"Yüklendi: {len(periodic_genes)} periyodik gen")

# 2. Ham veriyi oku
data_path = "/content/Spellman.csv"
df = pd.read_csv(data_path)

# 3. İlk sütundaki ORF isimlerine göre 1/0 etiketle
orf_column = df.columns[0]
df["label"] = df[orf_column].isin(periodic_genes).astype(int)

print("Etiket dağılımı:\n", df["label"].value_counts())

# 4. İşlenmiş veriyi kaydet
output_path = "Spellman_with_labels.csv"
df.to_csv(output_path, index=False)
print(f"Kaydedildi: {output_path}")
```
